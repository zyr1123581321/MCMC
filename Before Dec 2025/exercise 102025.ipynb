# Goal: Use Monte Carlo simulation to estimate the expectation of a random variable.


```python
import numpy as np
```

```python
def estimate_stats(N, f):
    X_k = np.random.randn(N)

    Y_k = f(X_k)

    mu_hat = np.mean(Y_k)
    sigma_hat = np.std(Y_k)

    return mu_hat, sigma_hat




```

```python
def f1(x):
    return x

def f2(x):
    return x**2

def f3(x):
    return x**3

def f4(x):
    return x**4

N_samples = 1000000

Funcs = [f1, f2, f3, f4]
true_mu = [0, 1, 0, 3]
true_variance = np.array([1,2,15,96])
true_std = np.sqrt(true_variance)


for i in range(4):
    mu_i, sig_i = estimate_stats(N_samples, Funcs[i])
    print(f"Case {i+1}: Y_k = X_k**{i+1} (N = 1000000)")
    print(f"True Mean:          {true_mu[i]:.6f}        True Standard Deviation:    {true_std[i]:.6f}")
    print(f"Estimated Mean:     {mu_i:.6f}      Estimated Standard Deviation: {sig_i:.6f}")
```

```python
def double_factorial(n):
    if n % 2 == 0:
        return None
    
    result = 1
    for i in range(n, 0, -2):
        result *= i
    return result
    
N = 5000000
n_max = 10

print(f"Testing E[X^(2n)] = (2n-1)!! using {N} samples")
print(f"{'n':<3} | {'Y = f(X)':<8} | {'Theoretical Mean':<18} | {'Estimated Mean':<22}")

for n in range(1, n_max +1):
    power = 2 * n

    f = lambda x, p=power: x**p

    true_mean =double_factorial(2 * n - 1)

    mu_hat, _ = estimate_stats(N, f)
    print(f"{n:<3} | {'X^' + str(power):<8} | {true_mean:<18} | {mu_hat:<22.6f}" )



```
